# Logistic Regression: hotel-booking cancellation

This notebook reproduces the current offline classifier. Target is hotel_bookings_updated_2024.csv ? is_canceled (1 = canceled, 0 = not canceled). The model is not deployed and does not estimate temperature, crowd, trip cost, or destination ratings.

The source has 119,390 transformed booking rows with arrival year 2024 only. Hotel/city mappings and data provenance are unverified. Results are diagnostic only.

````python
from pathlib import Path
import sys
import numpy as np
import pandas as pd
import sklearn
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer
from sklearn.dummy import DummyClassifier
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, classification_report, confusion_matrix,
    ConfusionMatrixDisplay, precision_recall_fscore_support,
    roc_auc_score, RocCurveDisplay, PrecisionRecallDisplay,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

SEED=42
def find_data_dir(required):
    for base in (Path.cwd(),*Path.cwd().parents):
        for candidate in (base/"Data",base/"Backend"/"Data"):
            if (candidate/required).is_file(): return candidate.resolve()
    raise FileNotFoundError(f"Cannot locate Backend/Data/{required}; open notebook within project.")
print("Python",sys.version.split()[0],"pandas",pd.__version__,"sklearn",sklearn.__version__,"NumPy",np.__version__)
DATA_DIR=find_data_dir("hotel_bookings_updated_2024.csv")
print("Located data directory:",DATA_DIR.name)
````

## Dataset overview and target definition

Positive class 1 means the booking row is marked canceled. reservation_status and reservation_status_date reveal the outcome and are excluded. The project-selected feature list is reproduced rather than using all columns. Missing country/children values are imputed inside the training pipeline; sparse agent/company are excluded.

````python
path=DATA_DIR/"hotel_bookings_updated_2024.csv"
if not path.is_file(): raise FileNotFoundError(f"Missing {path}")
raw=pd.read_csv(path,low_memory=False)
print("Shape",raw.shape,"\nColumns",raw.columns.tolist())
print("\nDtypes:\n",raw.dtypes.to_string())
print("\nMissing values:\n",raw.isna().sum()[raw.isna().sum()>0].to_string())
print("Exact duplicate rows",int(raw.duplicated().sum()))
print("Class counts",raw.is_canceled.value_counts(dropna=False).sort_index().to_dict())
print("Class proportions",raw.is_canceled.value_counts(normalize=True,dropna=False).sort_index().round(4).to_dict())
sample_cols=["hotel","is_canceled","lead_time","arrival_date_year","arrival_date_month","market_segment","adults","children"]
print("\nSample of target and selected features only:\n",raw[sample_cols].head(3).to_string(index=False))
print("\nNumeric descriptive statistics:\n",raw.describe().T.to_string())
````

## Date engineering, feature selection, and leakage checks

The existing pipeline maps arrival month names to numbers, creates the arrival date and derives arrival weekday. It uses 9 categorical and 15 numeric predictors. Explicit IDs, outcome fields and fields outside the project feature list are excluded. Because the source does not document the time at which every operational attribute is recorded, this is not a validated early-cancellation predictor.

````python
month_map={n:i for i,n in enumerate(["January","February","March","April","May","June","July","August","September","October","November","December"],1)}
data=raw.copy()
data["arrival_month_num"]=data.arrival_date_month.map(month_map)
data["arrival_date"]=pd.to_datetime({"year":data.arrival_date_year,"month":data.arrival_month_num,"day":data.arrival_date_day_of_month},errors="coerce")
data["arrival_dayofweek"]=data.arrival_date.dt.dayofweek
categorical=["hotel","arrival_date_month","meal","country","market_segment","distribution_channel","reserved_room_type","deposit_type","customer_type"]
numeric=["lead_time","arrival_date_week_number","arrival_date_day_of_month","arrival_dayofweek","stays_in_weekend_nights","stays_in_week_nights","adults","children","babies","is_repeated_guest","previous_cancellations","previous_bookings_not_canceled","days_in_waiting_list","total_of_special_requests","required_car_parking_spaces"]
features=categorical+numeric; target="is_canceled"
missing=sorted(set(features+[target,"arrival_date"])-set(data.columns))
if missing: raise ValueError(f"Missing required model columns: {missing}")
if data.arrival_date.isna().any(): raise ValueError("One or more arrival dates are invalid.")
if data[target].isna().any() or not set(data[target].unique()).issubset({0,1}): raise ValueError("is_canceled must be non-missing binary 0/1.")
excluded=["reservation_status","reservation_status_date","Booking_ID/Customer_ID","Customer_Name","agent","company","assigned_room_type","booking_changes","adr"]
print("Categorical",categorical,"\nNumeric",numeric,"\nExcluded",excluded)
print("Target included among predictors?",target in features)
print("Date span",data.arrival_date.min().date(),"to",data.arrival_date.max().date())
````

## Chronological split and train-only exploratory analysis

The current evaluation uses the earliest 80% of distinct arrival dates for training and latest 20% for testing; tied cutoff dates are wholly held out. No random shuffle or stratification is used for this time-dependent data. EDA plots use only training rows.

````python
dates=sorted(data.arrival_date.dropna().unique())
if len(dates)<5: raise ValueError("Not enough distinct dates for chronological holdout.")
cut=pd.Timestamp(dates[int(len(dates)*.8)])
train=data.loc[data.arrival_date<cut].copy(); test=data.loc[data.arrival_date>=cut].copy()
if train.empty or test.empty: raise ValueError("Chronological split has an empty partition.")
Xtr,ytr=train[features],train[target]; Xte,yte=test[features],test[target]
print("Test start",cut.date(),f"| train={len(train):,} | test={len(test):,}")
print("Train classes",ytr.value_counts().sort_index().to_dict())
print("Test classes",yte.value_counts().sort_index().to_dict())
fig,ax=plt.subplots(1,2,figsize=(10,4))
counts=ytr.value_counts().reindex([0,1],fill_value=0)
ax[0].bar(["Not canceled 0","Canceled 1"],counts.values,color=["#5b8b70","#bf7951"]); ax[0].set(title="Training class balance",ylabel="Booking rows"); ax[0].tick_params(axis="x",rotation=10)
bands=pd.cut(train.lead_time,bins=[-1,7,30,90,180,np.inf],labels=["0?7","8?30","31?90","91?180","181+"])
rate=train.assign(lead_band=bands).groupby("lead_band",observed=False)[target].mean()
ax[1].bar(rate.index.astype(str),rate.values,color="#3e7a96"); ax[1].set(title="Cancellation share by lead-time band",xlabel="Days (training)",ylabel="Share canceled")
fig.tight_layout(); plt.show()
````

## Preprocessing, majority baseline, and Logistic Regression

Categoricals: most-frequent imputation and one-hot encoding with unknown categories ignored. Numerics: median imputation and sparse-safe scaling. The ColumnTransformer is fit only on training rows.

Settings match the current code: max_iter=300, class_weight=balanced, solver=liblinear, random_state=42. Baseline is most-frequent class. Threshold is the default 0.5 and is not tuned against the test set. Class weighting changes precision/recall balance; probabilities are not claimed to be calibrated.

````python
prep=ColumnTransformer([
 ("categorical",Pipeline([("impute",SimpleImputer(strategy="most_frequent")),("onehot",OneHotEncoder(handle_unknown="ignore"))]),categorical),
 ("numeric",Pipeline([("impute",SimpleImputer(strategy="median")),("scale",StandardScaler(with_mean=False))]),numeric),
])
model=Pipeline([("preprocess",prep),("model",LogisticRegression(max_iter=300,class_weight="balanced",solver="liblinear",random_state=SEED))])
model.fit(Xtr,ytr); pred=model.predict(Xte)
classes=list(model.named_steps["model"].classes_); pos=classes.index(1) if 1 in classes else None
prob=model.predict_proba(Xte)[:,pos] if pos is not None else None
baseline=DummyClassifier(strategy="most_frequent").fit(np.zeros((len(ytr),1)),ytr)
bpred=baseline.predict(np.zeros((len(yte),1)))
bclasses=list(baseline.classes_); bpos=bclasses.index(1) if 1 in bclasses else None
bprob=baseline.predict_proba(np.zeros((len(yte),1)))[:,bpos] if bpos is not None else None
print("Positive class=1 (canceled); threshold=0.5; held-out predictions and probabilities created.")
````

## Fresh held-out metrics and baseline comparison

Precision, recall, F1 treat canceled=1 as positive. Confusion matrix order is [0,1]. ROC-AUC is calculated only when the test partition has both classes and probabilities for class 1; otherwise it is reported unavailable.

````python
def metrics(y,prediction,probability=None):
    precision,recall,f1,_=precision_recall_fscore_support(y,prediction,average="binary",pos_label=1,zero_division=0)
    auc=None
    if probability is not None and len(np.unique(y))==2:
        try: auc=roc_auc_score(y,probability)
        except ValueError: auc=None
    return {"Accuracy":accuracy_score(y,prediction),"Precision canceled=1":precision,"Recall canceled=1":recall,"F1 canceled=1":f1,"ROC-AUC canceled=1":auc}
result=pd.DataFrame({"Most-frequent baseline":metrics(yte,bpred,bprob),"Balanced Logistic Regression":metrics(yte,pred,prob)}).T
print("Dataset hotel_bookings_updated_2024.csv | target is_canceled | test start",cut.date())
print(f"Train={len(Xtr)} | test={len(Xte)} | train classes={ytr.value_counts().sort_index().to_dict()} | test classes={yte.value_counts().sort_index().to_dict()}")
print(result.round(4).to_string())
train_classes=list(model.named_steps["model"].classes_); train_pos=train_classes.index(1) if 1 in train_classes else None
train_prob=model.predict_proba(Xtr)[:,train_pos] if train_pos is not None else None
train_fit=metrics(ytr,model.predict(Xtr),train_prob)
print("\nOverfitting diagnostic: training fit versus held-out model metrics (training fit is optimistic):")
print(pd.DataFrame({"training fit":train_fit,"held-out":metrics(yte,pred,prob)}).T.round(4).to_string())
print("\nConfusion matrix rows=actual [0,1], cols=predicted [0,1]:\n",confusion_matrix(yte,pred,labels=[0,1]))
print("\nClassification report:\n",classification_report(yte,pred,labels=[0,1],target_names=["Not canceled 0","Canceled 1"],zero_division=0))
if prob is None or len(np.unique(yte))<2: print("ROC-AUC unavailable: both test classes and class-1 probabilities are required.")
````

## Confusion matrix, ROC and precision-recall charts

````python
ConfusionMatrixDisplay.from_predictions(yte,pred,labels=[0,1],display_labels=["Not canceled","Canceled"],cmap="Greens",values_format="d")
plt.title("Logistic Regression: chronological test confusion matrix"); plt.tight_layout(); plt.show()
if prob is not None and len(np.unique(yte))==2:
    fig,ax=plt.subplots(1,2,figsize=(11,4))
    RocCurveDisplay.from_predictions(yte,prob,ax=ax[0],name="Logistic Regression"); ax[0].set_title("ROC (held-out dates)")
    PrecisionRecallDisplay.from_predictions(yte,prob,ax=ax[1],name="Logistic Regression"); ax[1].set_title("Precision-recall (held-out dates)")
    fig.tight_layout(); plt.show()
else:
    print("ROC and precision-recall plots omitted: both classes/probabilities unavailable.")
````

## Interpretation and limitations

Class 1 means canceled. The 0.5 threshold is not tuned; balanced class weights alter precision/recall. A business threshold would require a separate validation set and explicit cost tradeoff.

The dataset covers arrival year 2024 only and has unverified transformed labels. The notebook compares training fit with held-out metrics as a simple overfitting check; training fit is optimistic. A chronological split within one year does not prove future-year generalization. Outcome fields are excluded, but the source does not establish that every other predictor is captured at a specific decision timestamp. Missing country/children values are imputed from training data; sparse agent/company and identifiers are not used.

This is an offline diagnostic, not a deployed cancellation predictor. Notebook metrics are recalculated from CSV; stored evaluation metrics are not copied. Website temperature/crowd are deterministic seasonal rules, not this classifier or any trained model.
